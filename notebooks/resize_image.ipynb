```python
import os
import cv2
import albumentations as A
import matplotlib.pyplot as plt
import random
```

```python
img = random.choice(os.listdir("../data/processed/images/val"))
img
```

```python
img_path = os.path.join("../data/processed/images/val", img)
img_path
```

```python
def display_image(root):
    img = random.choice(os.listdir(root))
    img_path = os.path.join(root, img)

    transform = A.Compose([
        A.Resize(height=1024, width=1024, p=1.0),
    ])


    image = cv2.imread(img_path)
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    
    resized_image = transform(image=image)['image']

    plt.figure(figsize=(10, 5))
    plt.imshow(resized_image)
    plt.axis('off')
    plt.show()
```

```python
root = "../data/processed/images/val"
```

```python
display_image(root)
```
